```python
import pandas as pd
data = {
    "Age_Group": ["18-25","26-35","36-45","46-60"],
    "Income": [25000,40000,50000,60000]
}
data = pd.DataFrame(data)
data
```

```python
data.apply(list).to_dict()
```

```python
data.describe()
```

```python
summary = data.groupby("Age_Group")["Income"].agg(['mean', 'median', 'min', 'max', 'std'])
summary
```

```python
summary = data.groupby("Age_Group")["Income"].agg(['mean', 'median', 'min', 'max', 'std'])
summary.apply(list).to_dict()
```

```python
mean_income = data.groupby("Age_Group")["Income"].mean()
print(mean_income)
```

```python
setosa = data[data["Species"] == "Iris-setosa"]
print("Iris-setosa Statistics:\n")
print(setosa.describe())
```

```python
versicolor = data[data["Species"] == "Iris-versicolor"]
print("Iris-versicolor Statistics:\n")
print(versicolor.describe())
```

```python
for Species in data['Species'].unique():
    print(f"\nStatistics for {Species}:")
    stats=data[data['Species']==Species].describe()
    print(stats)
```
