```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag



```

```python
text = "Data analytics is the process of analyzing data to extract meaningful insights."
```

```python
import nltk
nltk.download('punkt_tab')

tokens = word_tokenize(text)
print("Tokens:", tokens)
```

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')

pos_tags = pos_tag(tokens)
print("\nPOS Tags:", pos_tags)
```

```python
import nltk
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))
filtered_tokens = [w for w in tokens if w.lower() not in stop_words]
print("\nAfter Stopword Removal:", filtered_tokens)
```

```python
ps = PorterStemmer()
stemmed = [ps.stem(w) for w in filtered_tokens]
print("\nStemmed Words:", stemmed)
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered_tokens]
print("\nLemmatized Words:", lemmatized)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
documents = [
    "Data analytics is important",
    "Analytics helps in decision making",
    "Data science and analytics are related"
]
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

# Convert to DataFrame for better view
import pandas as pd
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer.get_feature_names_out())

print(df_tfidf)
```
